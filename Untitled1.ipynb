```python
#[adventure, Fantacy, Action][{"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 28, "name": "Action"}]'
import ast

def convertTheGenre(obj):
    L=[]
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
```

```python
convertTheGenre('[{"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 28, "name": "Action"}]')
```

```python
movies.columns
```

```python
movies['genres'] = movies['genres'].apply(convertTheGenre)
```

```python
movies['keywords'] = movies['keywords'].apply(convertTheGenre)
```

```python
def convertTheCast(obj):
    L=[]
    count=0
    for i in ast.literal_eval(obj):
        if count != 3:
            L.append(i['name'])
            count +=1
        else:
            break
        
    return L
```

```python
movies['cast']=movies['cast'].apply(convertTheCast)
```

```python

def convertTheCrew(obj):
    L=[]
    for i in ast.literal_eval(obj):
        if i['job']== 'Director':
            L.append(i['name'])
    return L
```

```python
movies['crew']=movies['crew'].apply(convertTheCrew)
```

```python
#stemming
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
#porter stemmer is class
#ps is we created a object
```

```python
def stem(txt):
    y=[]
#we converted list into strg
    for i in txt.split():
        y.append(ps.stem(i))

    return " ".join(y)
    #now strg into list
```
